## Load Raw Datasets

### Data Model Interpretation

The data model connects four major business processes:

**Customer Demand**
Customers → Sales Orders → Sales Order Items

**Inventory Operations**
Products + Warehouses → Inventory → Inventory Transactions

**Order Fulfillment**
Sales Orders → Shipments

**Supply & Replenishment**
Suppliers + Products + Warehouses → Purchase Orders

Together, these relationships allow sales demand, backlog, inventory availability, supplier performance, and order fulfillment to be analyzed as an integrated business process.
