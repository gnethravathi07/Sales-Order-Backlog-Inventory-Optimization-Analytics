### Recommendation Framework

Each recommendation will be evaluated using:

4. **Priority** — Critical, High, Medium, or Low
5. **Business Impact** — Revenue, working capital, service level, inventory, or operational efficiency
6. **Responsible Function** — Which business team should own the action?
7. **KPI to Monitor** — How should progress be measured?
8. **Implementation Horizon** — Immediate, short-term, or medium-term

- Backlog reduction

## Step 5 — Build Executive Baseline KPI Summary

Create a consolidated executive-level baseline of the most important business KPIs.

This establishes the current operational position before recommendations are prioritized.

The baseline covers:

- Sales and order volume
- Fulfillment and backlog
- Inventory exposure
- Replenishment pressure
- On-Time Delivery and OTIF
- Procurement performance
- Open purchase-order exposure

Only chronology-valid shipment records are used for delivery KPIs.

```python
# --------------------------------------------------
# Sales / Fulfillment / Backlog KPIs
# --------------------------------------------------

total_orders = sales_orders["order_id"].nunique()

total_sales_value = sales_orders["order_value"].sum()

total_ordered_qty = sales_orders["ordered_qty"].sum()

total_fulfilled_qty = sales_orders["fulfilled_qty"].sum()

total_backlog_qty = sales_orders["backlog_qty"].sum()

total_backlog_value = sales_orders["backlog_value"].sum()


overall_fill_rate_pct = (
    total_fulfilled_qty
    / total_ordered_qty
    * 100
)


backlog_qty_rate_pct = (
    total_backlog_qty
    / total_ordered_qty
    * 100
)


orders_with_backlog = (
    sales_orders["backlog_qty"] > 0
).sum()


backlog_order_rate_pct = (
    orders_with_backlog
    / total_orders
    * 100
)


# --------------------------------------------------
# Inventory KPIs
# --------------------------------------------------

total_inventory_value = (
    inventory["inventory_value"].sum()
)

reorder_required_records = (
    inventory["reorder_required_flag"] == "Y"
).sum()

reorder_required_rate_pct = (
    reorder_required_records
    / len(inventory)
    * 100
)

stockout_records = (
    inventory["stock_status"] == "Stockout"
).sum()

overstock_records = (
    inventory["stock_status"] == "Overstock"
).sum()


# --------------------------------------------------
# Valid delivery KPIs
# --------------------------------------------------

valid_shipments = (
    shipments[
        shipments["chronology_valid_flag"] == "Y"
    ]
    .copy()
)


valid_on_time_rate_pct = (
    (
        valid_shipments["on_time_flag_clean"]
        == "Y"
    ).mean()
    * 100
)


valid_otif_rate_pct = (
    (
        valid_shipments["otif_flag_clean"]
        == "Y"
    ).mean()
    * 100
)


avg_delivery_delay_days = (
    valid_shipments[
        "delivery_delay_days_clean"
    ].mean()
)


# --------------------------------------------------
# Procurement KPIs
# --------------------------------------------------

valid_po_delivery = (
    purchase_orders[
        purchase_orders[
            "supplier_delay_days"
        ].notna()
    ]
    .copy()
)


procurement_on_time_rate_pct = (
    (
        valid_po_delivery[
            "supplier_delay_days"
        ] <= 0
    ).mean()
    * 100
)


avg_supplier_delay_days = (
    valid_po_delivery[
        "supplier_delay_days"
    ].mean()
)


open_purchase_orders = (
    purchase_orders[
        "po_status"
    ] == "Open"
).sum()


open_po_value = (
    purchase_orders.loc[
        purchase_orders["po_status"] == "Open",
        "po_value"
    ].sum()
)


# --------------------------------------------------
# Executive KPI table
# --------------------------------------------------

executive_baseline = pd.DataFrame(
    [
        ["Total Orders", total_orders, "Orders"],
        ["Total Sales Value", total_sales_value, "Currency"],
        ["Overall Quantity Fill Rate", overall_fill_rate_pct, "%"],
        ["Orders With Backlog", orders_with_backlog, "Orders"],
        ["Backlog Order Rate", backlog_order_rate_pct, "%"],
        ["Backlog Quantity", total_backlog_qty, "Units"],
        ["Backlog Quantity Rate", backlog_qty_rate_pct, "%"],
        ["Backlog Value", total_backlog_value, "Currency"],
        ["Inventory Value", total_inventory_value, "Currency"],
        ["Reorder Required Records", reorder_required_records, "Records"],
        ["Reorder Required Rate", reorder_required_rate_pct, "%"],
        ["Stockout Records", stockout_records, "Records"],
        ["Overstock Records", overstock_records, "Records"],
        ["Valid On-Time Delivery Rate", valid_on_time_rate_pct, "%"],
        ["Valid OTIF Rate", valid_otif_rate_pct, "%"],
        ["Average Delivery Delay", avg_delivery_delay_days, "Days"],
        ["Procurement On-Time Rate", procurement_on_time_rate_pct, "%"],
        ["Average Supplier Delay", avg_supplier_delay_days, "Days"],
        ["Open Purchase Orders", open_purchase_orders, "POs"],
        ["Open PO Value", open_po_value, "Currency"]
    ],
    columns=[
        "kpi",
        "value",
        "unit"
    ]
)


print("EXECUTIVE BASELINE KPI SUMMARY\n")


display_baseline = executive_baseline.copy()

display_baseline["value"] = (
    display_baseline["value"]
    .map(
        lambda x:
        f"{x:,.2f}"
        if isinstance(x, (float, np.floating))
        else f"{x:,}"
    )
)


print(
    display_baseline.to_string(
        index=False
    )
)
```

## Step 6 — Backlog Reduction Priority Analysis

Evaluate backlog exposure across order-priority levels.

The objective is to identify where backlog reduction efforts should be concentrated based on:

- Total orders
- Backlogged orders
- Ordered quantity
- Backlog quantity
- Backlog rate
- Backlog value
- Share of total backlog value

This analysis will provide the evidence for the first management recommendation.

```python
# Backlog exposure by order priority

priority_backlog = (
    sales_orders
    .groupby(
        "priority",
        as_index=False
    )
    .agg(
        total_orders=("order_id", "nunique"),
        backlogged_orders=(
            "backlog_qty",
            lambda x: (x > 0).sum()
        ),
        ordered_qty=("ordered_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
)


# Backlog rate

priority_backlog["backlog_rate_pct"] = (
    priority_backlog["backlog_qty"]
    / priority_backlog["ordered_qty"]
    * 100
)


# Share of total backlog value

priority_backlog["backlog_value_share_pct"] = (
    priority_backlog["backlog_value"]
    / priority_backlog["backlog_value"].sum()
    * 100
)


# Sort by backlog value

priority_backlog = (
    priority_backlog
    .sort_values(
        "backlog_value",
        ascending=False
    )
    .reset_index(drop=True)
)


print("BACKLOG EXPOSURE BY ORDER PRIORITY\n")

print(
    priority_backlog.to_string(
        index=False,
        formatters={
            "backlog_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_value_share_pct":
                lambda x: f"{x:.2f}%",

            "backlog_value":
                lambda x: f"{x:,.2f}"
        }
    )
)
```

## Step 7 — High-Demand Product Backlog Risk

Evaluate backlog exposure according to product demand intensity.

Products will be divided into demand quartiles using their total ordered quantity.

The analysis will compare:

- Number of products
- Ordered quantity
- Backlog quantity
- Backlog rate
- Backlog value
- Share of total backlog quantity
- Share of total backlog value

This identifies whether inventory and fulfillment interventions should be concentrated on high-demand products rather than applied uniformly across the entire product portfolio.

```python
# Product-level demand and backlog summary

product_demand_backlog = (
    sales_order_items
    .groupby(
        "product_id",
        as_index=False
    )
    .agg(
        ordered_qty=("ordered_qty", "sum"),
        fulfilled_qty=("fulfilled_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
)


# Create demand quartiles

product_demand_backlog["demand_quartile"] = pd.qcut(
    product_demand_backlog["ordered_qty"],
    q=4,
    labels=[
        "Q1 Lowest Demand",
        "Q2",
        "Q3",
        "Q4 Highest Demand"
    ]
)


# Summarize demand quartiles

demand_backlog_priority = (
    product_demand_backlog
    .groupby(
        "demand_quartile",
        observed=True
    )
    .agg(
        products=("product_id", "nunique"),
        ordered_qty=("ordered_qty", "sum"),
        fulfilled_qty=("fulfilled_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
    .reset_index()
)


# Calculate performance metrics

demand_backlog_priority["backlog_rate_pct"] = (
    demand_backlog_priority["backlog_qty"]
    / demand_backlog_priority["ordered_qty"]
    * 100
)


demand_backlog_priority["backlog_qty_share_pct"] = (
    demand_backlog_priority["backlog_qty"]
    / demand_backlog_priority["backlog_qty"].sum()
    * 100
)


demand_backlog_priority["backlog_value_share_pct"] = (
    demand_backlog_priority["backlog_value"]
    / demand_backlog_priority["backlog_value"].sum()
    * 100
)


print("HIGH-DEMAND PRODUCT BACKLOG RISK\n")


print(
    demand_backlog_priority.to_string(
        index=False,
        formatters={
            "backlog_value":
                lambda x: f"{x:,.2f}",

            "backlog_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_qty_share_pct":
                lambda x: f"{x:.2f}%",

            "backlog_value_share_pct":
                lambda x: f"{x:.2f}%"
        }
    )
)
```

## Step 8 — High-Value Product Commercial Risk

Evaluate backlog exposure across product sales-value quartiles.

Products will be segmented according to total Net Order Value to determine whether commercially important products carry disproportionate backlog exposure.

The analysis will measure:

- Product count
- Sales value
- Share of total sales
- Backlog quantity
- Backlog value
- Backlog rate
- Share of total backlog value

This provides evidence for protecting high-value products through stronger inventory allocation, replenishment, and fulfillment controls.

```python
# Product-level commercial exposure

product_value_backlog = (
    sales_order_items
    .groupby(
        "product_id",
        as_index=False
    )
    .agg(
        sales_value=("net_order_value", "sum"),
        ordered_qty=("ordered_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
)


# Create sales-value quartiles

product_value_backlog["sales_value_quartile"] = pd.qcut(
    product_value_backlog["sales_value"],
    q=4,
    labels=[
        "Q1 Lowest Value",
        "Q2",
        "Q3",
        "Q4 Highest Value"
    ]
)


# Summarize quartiles

commercial_risk_summary = (
    product_value_backlog
    .groupby(
        "sales_value_quartile",
        observed=True
    )
    .agg(
        products=("product_id", "nunique"),
        sales_value=("sales_value", "sum"),
        ordered_qty=("ordered_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
    .reset_index()
)


commercial_risk_summary["sales_share_pct"] = (
    commercial_risk_summary["sales_value"]
    / commercial_risk_summary["sales_value"].sum()
    * 100
)


commercial_risk_summary["backlog_rate_pct"] = (
    commercial_risk_summary["backlog_qty"]
    / commercial_risk_summary["ordered_qty"]
    * 100
)


commercial_risk_summary["backlog_value_share_pct"] = (
    commercial_risk_summary["backlog_value"]
    / commercial_risk_summary["backlog_value"].sum()
    * 100
)


print("HIGH-VALUE PRODUCT COMMERCIAL RISK\n")


print(
    commercial_risk_summary.to_string(
        index=False,
        formatters={
            "sales_value":
                lambda x: f"{x:,.2f}",

            "backlog_value":
                lambda x: f"{x:,.2f}",

            "sales_share_pct":
                lambda x: f"{x:.2f}%",

            "backlog_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_value_share_pct":
                lambda x: f"{x:.2f}%"
        }
    )
)
```

## Step 9 — Inventory Replenishment Risk

Evaluate current inventory positions requiring replenishment.

The analysis will compare inventory records that require reorder against those that do not based on:

- Product-Warehouse record count
- Available quantity
- Inventory value
- Average inventory days on hand
- Stockout records
- Share of inventory records
- Share of inventory value

This provides the evidence for targeted replenishment rather than applying the same inventory policy across every Product-Warehouse combination.

```python
# Inventory replenishment risk summary

replenishment_risk = (
    inventory
    .groupby(
        "reorder_required_flag",
        as_index=False
    )
    .agg(
        records=("product_id", "size"),
        products=("product_id", "nunique"),
        available_qty=("available_qty", "sum"),
        inventory_value=("inventory_value", "sum"),
        avg_inventory_doh=(
            "inventory_days_on_hand",
            "mean"
        ),
        stockout_records=(
            "stock_status",
            lambda x: (x == "Stockout").sum()
        )
    )
)


# Record share

replenishment_risk["record_share_pct"] = (
    replenishment_risk["records"]
    / len(inventory)
    * 100
)


# Inventory value share

replenishment_risk["inventory_value_share_pct"] = (
    replenishment_risk["inventory_value"]
    / inventory["inventory_value"].sum()
    * 100
)


# Business-friendly status

replenishment_risk["replenishment_status"] = (
    replenishment_risk[
        "reorder_required_flag"
    ].map(
        {
            "Y": "Reorder Required",
            "N": "No Reorder Required"
        }
    )
)


# Reorder columns

replenishment_risk = replenishment_risk[
    [
        "replenishment_status",
        "records",
        "products",
        "available_qty",
        "inventory_value",
        "avg_inventory_doh",
        "stockout_records",
        "record_share_pct",
        "inventory_value_share_pct"
    ]
]


print("INVENTORY REPLENISHMENT RISK\n")


print(
    replenishment_risk.to_string(
        index=False,
        formatters={
            "inventory_value":
                lambda x: f"{x:,.2f}",

            "avg_inventory_doh":
                lambda x: f"{x:.2f}",

            "record_share_pct":
                lambda x: f"{x:.2f}%",

            "inventory_value_share_pct":
                lambda x: f"{x:.2f}%"
        }
    )
)
```

## Step 10 — Excess Inventory Reduction Opportunity

Evaluate inventory tied up in overstock positions.

The analysis will quantify:

- Overstock Product-Warehouse records
- Products affected
- Available quantity
- Inventory value
- Average inventory days on hand
- Share of total inventory records
- Share of total inventory value

The objective is to identify working-capital reduction opportunities while keeping excess-inventory actions separate from replenishment decisions.

```python
# Separate overstock inventory

overstock_inventory = (
    inventory[
        inventory["stock_status"] == "Overstock"
    ]
    .copy()
)


# Overstock summary

overstock_summary = pd.DataFrame(
    {
        "metric": [
            "Overstock Records",
            "Products Affected",
            "Available Quantity",
            "Overstock Inventory Value",
            "Average Inventory DOH",
            "Share of Inventory Records",
            "Share of Inventory Value"
        ],

        "value": [
            len(overstock_inventory),

            overstock_inventory[
                "product_id"
            ].nunique(),

            overstock_inventory[
                "available_qty"
            ].sum(),

            overstock_inventory[
                "inventory_value"
            ].sum(),

            overstock_inventory[
                "inventory_days_on_hand"
            ].mean(),

            (
                len(overstock_inventory)
                / len(inventory)
                * 100
            ),

            (
                overstock_inventory[
                    "inventory_value"
                ].sum()
                / inventory[
                    "inventory_value"
                ].sum()
                * 100
            )
        ]
    }
)


print("EXCESS INVENTORY REDUCTION OPPORTUNITY\n")


for _, row in overstock_summary.iterrows():

    metric = row["metric"]
    value = row["value"]

    if metric in [
        "Overstock Inventory Value"
    ]:

        formatted_value = f"{value:,.2f}"

    elif metric in [
        "Average Inventory DOH",
        "Share of Inventory Records",
        "Share of Inventory Value"
    ]:

        formatted_value = f"{value:.2f}"

    else:

        formatted_value = f"{value:,.0f}"

    print(
        f"{metric:<32} : {formatted_value}"
    )
```

## Step 11 — Delivery & OTIF Improvement Priority

Evaluate valid shipment performance across order-priority levels.

Only shipments with valid chronology are included.

The analysis will compare:

- Shipment volume
- On-Time Delivery Rate
- Late Shipment Rate
- OTIF Rate
- Average Delivery Delay

This identifies whether high-priority customer orders are receiving the service levels expected from their urgency classification.

```python
# Merge valid shipments with order priority

priority_delivery = (
    valid_shipments[
        [
            "order_id",
            "delivery_delay_days_clean",
            "on_time_flag_clean",
            "otif_flag_clean"
        ]
    ]
    .merge(
        sales_orders[
            [
                "order_id",
                "priority"
            ]
        ],
        on="order_id",
        how="left",
        validate="one_to_one"
    )
)


# Summarize delivery performance by priority

priority_delivery_summary = (
    priority_delivery
    .groupby(
        "priority",
        as_index=False
    )
    .agg(
        shipments=("order_id", "count"),

        on_time_rate_pct=(
            "on_time_flag_clean",
            lambda x: (x == "Y").mean() * 100
        ),

        late_rate_pct=(
            "on_time_flag_clean",
            lambda x: (x == "N").mean() * 100
        ),

        otif_rate_pct=(
            "otif_flag_clean",
            lambda x: (x == "Y").mean() * 100
        ),

        avg_delivery_delay_days=(
            "delivery_delay_days_clean",
            "mean"
        )
    )
)


# Sort weakest OTIF first

priority_delivery_summary = (
    priority_delivery_summary
    .sort_values(
        "otif_rate_pct"
    )
    .reset_index(drop=True)
)


print("DELIVERY & OTIF PERFORMANCE BY PRIORITY\n")


print(
    priority_delivery_summary.to_string(
        index=False,
        formatters={
            "on_time_rate_pct":
                lambda x: f"{x:.2f}%",

            "late_rate_pct":
                lambda x: f"{x:.2f}%",

            "otif_rate_pct":
                lambda x: f"{x:.2f}%",

            "avg_delivery_delay_days":
                lambda x: f"{x:.2f}"
        }
    )
)
```

## Step 12 — Fulfillment Impact on Delivery Performance

Evaluate delivery performance according to whether an order was completely fulfilled before shipment.

Orders will be classified as:

- Fully Fulfilled
- Partially Fulfilled / Backlogged

The analysis compares:

- Order volume
- Average fill rate
- On-Time Delivery Rate
- OTIF Rate
- Average Delivery Delay

This provides evidence for improving order completeness before shipment as a service-level intervention.

```python
# Merge valid shipments with fulfillment information

fulfillment_delivery_recommendation = (
    valid_shipments[
        [
            "order_id",
            "delivery_delay_days_clean",
            "on_time_flag_clean",
            "otif_flag_clean"
        ]
    ]
    .merge(
        sales_orders[
            [
                "order_id",
                "order_fill_rate_pct",
                "backlog_qty"
            ]
        ],
        on="order_id",
        how="left",
        validate="one_to_one"
    )
)


# Classify fulfillment status

fulfillment_delivery_recommendation[
    "fulfillment_status"
] = np.where(
    fulfillment_delivery_recommendation[
        "backlog_qty"
    ] == 0,
    "Fully Fulfilled",
    "Partially Fulfilled / Backlogged"
)


# Summarize performance

fulfillment_delivery_summary = (
    fulfillment_delivery_recommendation
    .groupby(
        "fulfillment_status",
        as_index=False
    )
    .agg(
        orders=("order_id", "count"),

        avg_fill_rate_pct=(
            "order_fill_rate_pct",
            "mean"
        ),

        on_time_rate_pct=(
            "on_time_flag_clean",
            lambda x: (x == "Y").mean() * 100
        ),

        otif_rate_pct=(
            "otif_flag_clean",
            lambda x: (x == "Y").mean() * 100
        ),

        avg_delivery_delay_days=(
            "delivery_delay_days_clean",
            "mean"
        )
    )
)


print("FULFILLMENT IMPACT ON DELIVERY PERFORMANCE\n")


print(
    fulfillment_delivery_summary.to_string(
        index=False,
        formatters={
            "avg_fill_rate_pct":
                lambda x: f"{x:.2f}%",

            "on_time_rate_pct":
                lambda x: f"{x:.2f}%",

            "otif_rate_pct":
                lambda x: f"{x:.2f}%",

            "avg_delivery_delay_days":
                lambda x: f"{x:.2f}"
        }
    )
)
```

## Step 13 — Supplier Performance Intervention Priority

Evaluate supplier performance using both operational reliability and procurement exposure.

Supplier performance will be summarized using:

- Purchase-order volume
- Procurement value
- Reliability Factor
- On-Time Procurement Rate
- Average Supplier Delay
- Delayed Purchase Orders

The objective is to identify suppliers where poor delivery performance overlaps with meaningful procurement exposure.

This supports targeted supplier-performance management rather than applying the same intervention to every supplier.

```python
# Merge purchase orders with supplier information

supplier_po_analysis = (
    valid_po_delivery
    .merge(
        suppliers[
            [
                "supplier_id",
                "reliability_factor"
            ]
        ],
        on="supplier_id",
        how="left",
        validate="many_to_one"
    )
)


# Supplier-level performance

supplier_performance = (
    supplier_po_analysis
    .groupby(
        "supplier_id",
        as_index=False
    )
    .agg(
        purchase_orders=("po_id", "count"),

        procurement_value=("po_value", "sum"),

        reliability_factor=(
            "reliability_factor",
            "first"
        ),

        on_time_rate_pct=(
            "supplier_delay_days",
            lambda x: (x <= 0).mean() * 100
        ),

        avg_supplier_delay_days=(
            "supplier_delay_days",
            "mean"
        ),

        delayed_pos=(
            "supplier_delay_days",
            lambda x: (x > 0).sum()
        )
    )
)


# Procurement exposure share

supplier_performance[
    "procurement_value_share_pct"
] = (
    supplier_performance[
        "procurement_value"
    ]
    / supplier_performance[
        "procurement_value"
    ].sum()
    * 100
)


# Define intervention candidates:
# below 60% on-time OR reliability below 0.70

supplier_performance[
    "intervention_required"
] = np.where(
    (
        supplier_performance[
            "on_time_rate_pct"
        ] < 60
    )
    |
    (
        supplier_performance[
            "reliability_factor"
        ] < 0.70
    ),
    "Y",
    "N"
)


supplier_intervention = (
    supplier_performance[
        supplier_performance[
            "intervention_required"
        ] == "Y"
    ]
    .sort_values(
        "procurement_value",
        ascending=False
    )
    .reset_index(drop=True)
)


print("SUPPLIER PERFORMANCE INTERVENTION PRIORITY\n")

print(
    "Suppliers Requiring Intervention :",
    len(supplier_intervention)
)

print(
    "Procurement Exposure             :",
    f"{supplier_intervention['procurement_value'].sum():,.2f}"
)

print(
    "Exposure Share                   :",
    f"{supplier_intervention['procurement_value_share_pct'].sum():.2f}%"
)


print("\nTOP 10 INTERVENTION PRIORITIES\n")


print(
    supplier_intervention[
        [
            "supplier_id",
            "purchase_orders",
            "procurement_value",
            "reliability_factor",
            "on_time_rate_pct",
            "avg_supplier_delay_days",
            "delayed_pos"
        ]
    ]
    .head(10)
    .to_string(
        index=False,
        formatters={
            "procurement_value":
                lambda x: f"{x:,.2f}",

            "reliability_factor":
                lambda x: f"{x:.3f}",

            "on_time_rate_pct":
                lambda x: f"{x:.2f}%",

            "avg_supplier_delay_days":
                lambda x: f"{x:.2f}"
        }
    )
)
```

## Step 14 — Warehouse Intervention Priority

Evaluate warehouse performance across multiple operational dimensions.

The analysis integrates:

- Reorder-required rate
- Stockout records
- Quantity fill rate
- Backlog rate
- Backlog value
- On-Time Delivery Rate
- OTIF Rate

The objective is to identify warehouses where multiple operational weaknesses overlap.

This supports targeted warehouse intervention rather than evaluating inventory, fulfillment, and delivery KPIs independently.

```python
# --------------------------------------------------
# Inventory performance by warehouse
# --------------------------------------------------

warehouse_inventory = (
    inventory
    .groupby(
        "warehouse_id",
        as_index=False
    )
    .agg(
        inventory_records=("product_id", "size"),

        reorder_required_records=(
            "reorder_required_flag",
            lambda x: (x == "Y").sum()
        ),

        stockout_records=(
            "stock_status",
            lambda x: (x == "Stockout").sum()
        ),

        inventory_value=(
            "inventory_value",
            "sum"
        )
    )
)


warehouse_inventory["reorder_rate_pct"] = (
    warehouse_inventory[
        "reorder_required_records"
    ]
    / warehouse_inventory[
        "inventory_records"
    ]
    * 100
)


# --------------------------------------------------
# Fulfillment performance by warehouse
# --------------------------------------------------

warehouse_fulfillment = (
    sales_orders
    .groupby(
        "warehouse_id",
        as_index=False
    )
    .agg(
        orders=("order_id", "nunique"),
        ordered_qty=("ordered_qty", "sum"),
        fulfilled_qty=("fulfilled_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
)


warehouse_fulfillment["fill_rate_pct"] = (
    warehouse_fulfillment[
        "fulfilled_qty"
    ]
    / warehouse_fulfillment[
        "ordered_qty"
    ]
    * 100
)


warehouse_fulfillment["backlog_rate_pct"] = (
    warehouse_fulfillment[
        "backlog_qty"
    ]
    / warehouse_fulfillment[
        "ordered_qty"
    ]
    * 100
)


# --------------------------------------------------
# Delivery performance by warehouse
# --------------------------------------------------

warehouse_delivery = (
    valid_shipments
    .groupby(
        "warehouse_id",
        as_index=False
    )
    .agg(
        valid_shipments=("shipment_id", "count"),

        on_time_rate_pct=(
            "on_time_flag_clean",
            lambda x: (x == "Y").mean() * 100
        ),

        otif_rate_pct=(
            "otif_flag_clean",
            lambda x: (x == "Y").mean() * 100
        )
    )
)


# --------------------------------------------------
# Combine warehouse KPIs
# --------------------------------------------------

warehouse_intervention = (
    warehouses[
        [
            "warehouse_id",
            "warehouse_name"
        ]
    ]
    .merge(
        warehouse_inventory,
        on="warehouse_id",
        how="left"
    )
    .merge(
        warehouse_fulfillment,
        on="warehouse_id",
        how="left"
    )
    .merge(
        warehouse_delivery,
        on="warehouse_id",
        how="left"
    )
)


# Display management-focused fields

warehouse_intervention_display = (
    warehouse_intervention[
        [
            "warehouse_id",
            "warehouse_name",
            "reorder_rate_pct",
            "stockout_records",
            "fill_rate_pct",
            "backlog_rate_pct",
            "backlog_value",
            "on_time_rate_pct",
            "otif_rate_pct"
        ]
    ]
    .sort_values(
        [
            "reorder_rate_pct",
            "backlog_rate_pct"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)


print("WAREHOUSE INTERVENTION PRIORITY\n")


print(
    warehouse_intervention_display.to_string(
        index=False,
        formatters={
            "reorder_rate_pct":
                lambda x: f"{x:.2f}%",

            "fill_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_value":
                lambda x: f"{x:,.2f}",

            "on_time_rate_pct":
                lambda x: f"{x:.2f}%",

            "otif_rate_pct":
                lambda x: f"{x:.2f}%"
        }
    )
)
```

## Step 15 — Category-Level Intervention Priority

Evaluate product categories across commercial and operational performance.

The analysis integrates:

- Sales value
- Backlog value
- Quantity fill rate
- Backlog rate
- Inventory value
- Reorder-required rate
- Stockout records

The objective is to distinguish between:

- Categories with weak fulfillment performance
- Categories with high commercial exposure
- Categories with greater replenishment pressure

This allows category-specific recommendations instead of applying a uniform inventory and fulfillment strategy.

```python
# --------------------------------------------------
# Sales / backlog by category
# --------------------------------------------------

category_sales = (
    sales_order_items
    .merge(
        products[
            [
                "product_id",
                "category"
            ]
        ],
        on="product_id",
        how="left",
        validate="many_to_one"
    )
)


category_fulfillment = (
    category_sales
    .groupby(
        "category",
        as_index=False
    )
    .agg(
        sales_value=("net_order_value", "sum"),
        ordered_qty=("ordered_qty", "sum"),
        fulfilled_qty=("fulfilled_qty", "sum"),
        backlog_qty=("backlog_qty", "sum"),
        backlog_value=("backlog_value", "sum")
    )
)


category_fulfillment["fill_rate_pct"] = (
    category_fulfillment["fulfilled_qty"]
    / category_fulfillment["ordered_qty"]
    * 100
)


category_fulfillment["backlog_rate_pct"] = (
    category_fulfillment["backlog_qty"]
    / category_fulfillment["ordered_qty"]
    * 100
)


category_fulfillment["sales_share_pct"] = (
    category_fulfillment["sales_value"]
    / category_fulfillment["sales_value"].sum()
    * 100
)


category_fulfillment["backlog_value_share_pct"] = (
    category_fulfillment["backlog_value"]
    / category_fulfillment["backlog_value"].sum()
    * 100
)


# --------------------------------------------------
# Inventory by category
# --------------------------------------------------

category_inventory_base = (
    inventory
    .merge(
        products[
            [
                "product_id",
                "category"
            ]
        ],
        on="product_id",
        how="left",
        validate="many_to_one"
    )
)


category_inventory = (
    category_inventory_base
    .groupby(
        "category",
        as_index=False
    )
    .agg(
        inventory_records=("product_id", "size"),

        inventory_value=("inventory_value", "sum"),

        reorder_required_records=(
            "reorder_required_flag",
            lambda x: (x == "Y").sum()
        ),

        stockout_records=(
            "stock_status",
            lambda x: (x == "Stockout").sum()
        )
    )
)


category_inventory["reorder_rate_pct"] = (
    category_inventory["reorder_required_records"]
    / category_inventory["inventory_records"]
    * 100
)


# --------------------------------------------------
# Combine category KPIs
# --------------------------------------------------

category_intervention = (
    category_fulfillment
    .merge(
        category_inventory,
        on="category",
        how="left"
    )
    .sort_values(
        "backlog_rate_pct",
        ascending=False
    )
    .reset_index(drop=True)
)


print("CATEGORY-LEVEL INTERVENTION PRIORITY\n")


print(
    category_intervention[
        [
            "category",
            "sales_value",
            "sales_share_pct",
            "fill_rate_pct",
            "backlog_rate_pct",
            "backlog_value",
            "backlog_value_share_pct",
            "inventory_value",
            "reorder_rate_pct",
            "stockout_records"
        ]
    ]
    .to_string(
        index=False,
        formatters={
            "sales_value":
                lambda x: f"{x:,.2f}",

            "sales_share_pct":
                lambda x: f"{x:.2f}%",

            "fill_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_rate_pct":
                lambda x: f"{x:.2f}%",

            "backlog_value":
                lambda x: f"{x:,.2f}",

            "backlog_value_share_pct":
                lambda x: f"{x:.2f}%",

            "inventory_value":
                lambda x: f"{x:,.2f}",

            "reorder_rate_pct":
                lambda x: f"{x:.2f}%"
        }
    )
)
```

## Step 16 — Inbound PO Coverage for Replenishment Risk

Evaluate whether Product-Warehouse combinations requiring replenishment already have incoming purchase-order quantities.

The analysis will measure:

- Reorder-required records
- Records with outstanding inbound PO quantity
- Records without inbound coverage
- Inbound coverage rate
- Total outstanding PO quantity
- Available quantity and inventory days on hand

This prevents unnecessary purchasing by distinguishing between inventory shortages already being replenished and shortages that still require procurement action.

Inbound PO presence represents procurement response and does not necessarily mean that the full replenishment requirement is covered.

```python
# --------------------------------------------------
# Outstanding PO quantity by Product-Warehouse
# --------------------------------------------------

po_inbound = (
    purchase_orders
    .assign(
        outstanding_po_qty=lambda df:
            (
                df["ordered_qty"]
                - df["received_qty"]
            ).clip(lower=0)
    )
    .groupby(
        [
            "product_id",
            "warehouse_id"
        ],
        as_index=False
    )
    .agg(
        outstanding_po_qty=(
            "outstanding_po_qty",
            "sum"
        )
    )
)


# --------------------------------------------------
# Merge inbound supply with inventory
# --------------------------------------------------

inventory_inbound = (
    inventory
    .merge(
        po_inbound,
        on=[
            "product_id",
            "warehouse_id"
        ],
        how="left"
    )
)


inventory_inbound[
    "outstanding_po_qty"
] = (
    inventory_inbound[
        "outstanding_po_qty"
    ]
    .fillna(0)
)


inventory_inbound[
    "inbound_po_flag"
] = np.where(
    inventory_inbound[
        "outstanding_po_qty"
    ] > 0,
    "Y",
    "N"
)


# --------------------------------------------------
# Focus on reorder-required records
# --------------------------------------------------

reorder_inbound = (
    inventory_inbound[
        inventory_inbound[
            "reorder_required_flag"
        ] == "Y"
    ]
    .copy()
)


reorder_records = len(
    reorder_inbound
)

records_with_inbound = (
    reorder_inbound[
        "inbound_po_flag"
    ] == "Y"
).sum()

records_without_inbound = (
    reorder_inbound[
        "inbound_po_flag"
    ] == "N"
).sum()

inbound_coverage_rate_pct = (
    records_with_inbound
    / reorder_records
    * 100
)


print("INBOUND PO COVERAGE FOR REPLENISHMENT RISK\n")

print(
    f"{'Reorder Required Records':<38} : "
    f"{reorder_records:,}"
)

print(
    f"{'Records With Inbound PO':<38} : "
    f"{records_with_inbound:,}"
)

print(
    f"{'Records Without Inbound PO':<38} : "
    f"{records_without_inbound:,}"
)

print(
    f"{'Inbound Coverage Rate':<38} : "
    f"{inbound_coverage_rate_pct:.2f}%"
)

print(
    f"{'Outstanding PO Quantity':<38} : "
    f"{reorder_inbound['outstanding_po_qty'].sum():,.0f}"
)

print(
    f"{'Average Available Quantity':<38} : "
    f"{reorder_inbound['available_qty'].mean():.2f}"
)

print(
    f"{'Average Inventory DOH':<38} : "
    f"{reorder_inbound['inventory_days_on_hand'].mean():.2f}"
)
```

## Step 18 — Recommendation Prioritization Score

Create a structured prioritization score for the management recommendations.

The score combines:

- Priority severity
- Business-impact breadth
- Implementation urgency

The scoring framework is used to organize recommendations for management attention.

It is a decision-support prioritization mechanism rather than a prediction of financial return or guaranteed business impact.

```python
# --------------------------------------------------
# Recommendation prioritization scoring
# --------------------------------------------------

priority_score_map = {
    "Critical": 4,
    "High": 3,
    "Medium": 2,
    "Low": 1
}


horizon_score_map = {
    "Immediate": 3,
    "Short-Term": 2,
    "Medium-Term": 1
}


# Functions containing multiple business areas
# receive a slightly higher cross-functional score.

recommendation_scoring = (
    recommendation_matrix.copy()
)


recommendation_scoring[
    "priority_score"
] = (
    recommendation_scoring[
        "priority"
    ].map(priority_score_map)
)


recommendation_scoring[
    "urgency_score"
] = (
    recommendation_scoring[
        "implementation_horizon"
    ].map(horizon_score_map)
)


recommendation_scoring[
    "cross_functional_score"
] = (
    recommendation_scoring[
        "responsible_function"
    ]
    .str.contains(r"\+")
    .map({
        True: 2,
        False: 1
    })
)


# Maximum score = 9

recommendation_scoring[
    "total_priority_score"
] = (
    recommendation_scoring[
        "priority_score"
    ]
    +
    recommendation_scoring[
        "urgency_score"
    ]
    +
    recommendation_scoring[
        "cross_functional_score"
    ]
)


# Management tier

recommendation_scoring[
    "management_tier"
] = pd.cut(
    recommendation_scoring[
        "total_priority_score"
    ],
    bins=[
        0,
        5,
        7,
        9
    ],
    labels=[
        "Tier 3 - Monitor",
        "Tier 2 - High",
        "Tier 1 - Executive Focus"
    ],
    include_lowest=True
)


recommendation_scoring = (
    recommendation_scoring
    .sort_values(
        [
            "total_priority_score",
            "recommendation_id"
        ],
        ascending=[
            False,
            True
        ]
    )
    .reset_index(drop=True)
)


print("RECOMMENDATION PRIORITIZATION SCORE\n")


print(
    recommendation_scoring[
        [
            "recommendation_id",
            "business_issue",
            "priority",
            "implementation_horizon",
            "total_priority_score",
            "management_tier"
        ]
    ].to_string(index=False)
)


print("\nMANAGEMENT TIER DISTRIBUTION\n")

print(
    recommendation_scoring[
        "management_tier"
    ]
    .value_counts(sort=False)
    .to_string()
)
```
